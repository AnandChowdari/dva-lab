### 10. Jupyter Magic Commands
`dva-lab` automatically registers IPython magics when imported!
- `%load_code 3` : Replaces cell content with the source code of Experiment 3.
- `%show_experiment 3` : Displays the full documentation.
- `%list_experiments` : Displays the catalog.
- `%run_experiment 3` : Executes the experiment inline.
